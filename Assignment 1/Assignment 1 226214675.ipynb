```python
import hashlib
import time
from pathlib import Path
```

```python
#Question 1 digesting a string
def sha256_string(s: str) -> str:
    return hashlib.sha256(s.encode('utf-8')).hexdigest()


print(sha256_string("financial"))
```

```python
def sha256_file(path: str) -> str:
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(8192), b""):
            h.update(chunk)
    return h.hexdigest()

# Example usage: Hashing a Desktop file
desktop_file = r"C:\Users\liman\Downloads\Motivation.docx"
file_hash = sha256_file(desktop_file)

print(f"SHA-256 Digest: {file_hash}")
```

```python
#Question 2 
#Showing the avalanche effect
def sha256_string(s: str) -> str:
    return hashlib.sha256(s.encode('utf-8')).hexdigest()


print(sha256_string("financial"))
print(sha256_string("Financial"))
```

```python
#Question 3 
def proof_of_work(message: str, n: int) -> dict:
    target_prefix = "0" * n
    nonce = 0
    start_time = time.perf_counter()

    while True:
        candidate = f"{message}{nonce}"
        digest = hashlib.sha256(candidate.encode("utf-8")).hexdigest()
        
        if digest.startswith(target_prefix):
            elapsed_seconds = time.perf_counter() - start_time
            return {
                "nonce": nonce,
                "digest": digest,
                "attempts": nonce + 1,
                "seconds": elapsed_seconds
            }
        
        nonce += 1


# Print table header
print(f"{'n':<3} | {'Attempts':<10} | {'Time (s)':<10} | {'Digest Prefix (16 hex)':<18}")
print("-" * 52)

# trying PoW for n being  3 and 4
for n in (3, 4):
    r = proof_of_work("ledger-commit", n)
    print(f"{n:<3} | {r['attempts']:<10} | {r['seconds']:<10.4f}s | {r['digest'][:16]}")
```
